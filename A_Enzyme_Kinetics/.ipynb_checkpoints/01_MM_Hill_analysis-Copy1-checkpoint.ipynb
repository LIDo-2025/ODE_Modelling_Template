# Michaelis Menten & Hill Kinetics

Michaelis Menten (MM) and Hill kinetics in vitro (closed system). MM is nonlinear but monotonous. Hill with n>1 is nonlinear with a qualitative transition, the inflection point. Predict n=1 -> n=4 sigmoidal changes in open flow. Open flow and nonlinearity are the basic ingredients of every complex living systems. Concept: mathematical treatment of nonlinear functions.

## Motivation: Why These Equations Are Useful

In biological systems, the rate of a chemical reaction or a receptor-binding event is rarely a simple linear function of the concentration of a molecule. Instead, it rises steeply at first and then plateaus (saturates) because the enzyme or receptor has a finite number of binding sites.

The core need is a simple, interpretable equation that captures:

Saturation (rate → Vmax at high substrate/ligand concentration).

Cooperative effects where binding of one molecule makes subsequent binding easier or harder.

Half-maximal points (e.g., Km for enzymes, EC50 or Kd for receptors)

Without these equations, you would need to simulate every individual binding step (ordinary differential equations for each intermediate state), which becomes computationally expensive. They are reduced-order models that hide microscopic complexity behind a few phenomenological parameters.

## Historical Background

Early 20th century – Enzyme kinetics:

1903: Victor Henri observed that enzyme reaction rates saturate and proposed an equation with an intermediate complex.

1913: Leonor Michaelis and Maud Menten (a German physician and a Canadian physician/biochemist) re-derived and experimentally validated the equation v = Vmax*[S]/(Km+[S])

They assumed that substrate binding is fast equilibrium compared to the catalytic step. This explained why rate is linear at low [S] (first order) and constant at high [S] (zero order).


## The generalization – Cooperativity

1910: Archibald Hill, while studying oxygen binding to hemoglobin (which has 4 subunits), noticed that the binding curve was steeper than a Michaelis-Menten (non-hyperbolic, sigmoidal). He proposed an empirical equation: 

$$\theta = \frac{[L]^n"}{K_d + [L]^n}$$

 
where θ is fractional occupancy, [L] is ligand concentration, $n$ is the Hill coefficient, and 
$K_d$ is the dissociation constant (often written as $EC_{50}^$).

Why n=1 is special: When n=1, the Hill equation reduces exactly to the Michaelis-Menten form. This describes non-cooperative binding – each site acts independently. For enzymes, n=1 means the Michaelis-Menten equation works.

Interpretation of n:

n>1 : positive cooperativity (e.g., hemoglobin – binding first O₂ increases affinity for next).

n<1 : negative cooperativity (rare; binding first ligand reduces affinity).

Note: n is usually not equal to the number of binding sites – it’s an empirical slope measure.

## Use in Computational Modelling

Today, these equations are embedded in almost all multi-scale models of biological networks. Here are three concrete use cases:

1. Whole-Cell and Signalling Pathway Models (ODE-based)
In systems biology (e.g., modelling MAPK/ERK pathway, calcium dynamics, or gene regulation), thousands of reactions would be impractical. Instead: An enzyme-catalysed reaction is written as a contribution to the rate equation $\frac{dP}{dt}=V_{max}\cdot S / (K_m + S) $, ignoring the explicit enzyme-substrate complex.

A transcription factor binding to a promoter with cooperativity uses $\frac{dmRNA}{dt} = \beta \cdot [TF]^n / (K_{d}^{n} + [TF]^n ) $

Example tool: COPASI, SBML-compatible solvers, or MATLAB’s SimBiology.

2. Pharmacokinetics/Pharmacodynamics (PK/PD)

To model drug-receptor interaction *in vivo*:

- The Hill equation (often called *Emax model* in pharmacology) predicts effect 
*E* from drug concentration *C*:

$$E = E_0 + \frac{E_{max}\cdot C^n }{EC_{50}^{n} + C^n}$$

- Useful: For a new drug, you fit $EC_{50}$ and $n$ from dose-response data. If $n>1$, small dose increases cause large effect jumps (steep response) – important for safety margins.

3. Parameter Reduction in Computational Neuroscience

When modelling synaptic transmission or ion channels:

- The rate of neurotransmitter release as a function of calcium concentration often uses a Hill equation with n≈4 (e.g. cooperative Ca²⁺ binding to synaptotagmin).

This reduces a multi-step Ca²⁺-binding cascade to a single equation, making large network models (e.g., 1000-neuron circuits) computationally tractable.

4. Model Inference and Fitting (Data Science)
In single-cell RNA-seq or proteomics, the Hill equation is used as a transfer function between a regulator concentration and target gene expression. With Bayesian inference or non-linear least squares, you estimate $n$, $EC_{50}$, and $V_{max}$ from noisy data – revealing whether the regulatory step is switch-like (n≫1).


## A Caution for Modelling

The Hill equation is empirically derived and thermodynamically inconsistent for 
n>1 if derived from a simple mass-action binding scheme (except when n=1). 
For true mechanistic models with several binding steps, you would use the Adair equation or MWC (Monod-Wyman-Changeux) model. However, the Hill equation’s simplicity and good fit to many cooperative systems make it the workhorse of applied modelling.

Summary Table

| Feature                      | Michaelis-Menten (\( n=1 \))              | Hill (\( n \neq 1 \))                      |
|------------------------------|-------------------------------------------|--------------------------------------------|
| Curve shape                  | Hyperbolic                                | Sigmoidal (if \( n>1 \))                   |
| Biological meaning           | Non-cooperative enzyme / receptor         | Cooperative binding                        |
| Key parameter(s)             | \( K_m \) (affinity for substrate)        | \( EC_{50} \) and \( n \) (steepness)      |
| Modern use                   | Metabolic networks, single-substrate enzymes | Drug dose-response, gene regulation, ion channels |


We will first look at the equation in isolation, as a (static) rate law. Later this rate law can be re-used in dynamical models for simulations. 

# Illustration of Function

```python
from matplotlib.pyplot import figure

import numpy as np
```

## Rate function definition


```python
def menten_hill_kinetics(S, k_max, K_m, n):
    """
    Hill kinetics equation; reduces to Michaelis-Menten for n=1
    
    S:     substrate concentration (typically a range of values expected)
    k_max: maximum reaction rate (includes enzyme concentration)
    K_m:   Michaelis constant (fixed at 1)
    n:     Hill coefficient (exponent of cooperativity, positive real number)
    """

    rate = (k_max * S**n) / (K_m**n + S**n)
    
    return rate
```

Here, you specify a range of substrate concentration and the three constants needed to evaluate the function. The function call will return a range of velocities corresponding to the substrate concentrations. 

## Call of the rate function and plot


```python
# Parameters
K_m   = 1
k_max = 1

# Ranges for substrate and Hill coefficient
S_values = np.linspace(0,   3, 100)  
n_values = np.linspace(0.5, 4, 100)

# Meshgrid of S and n
S, N = np.meshgrid(S_values, n_values)

# Reaction rate and its gradient
rate_landscape = np.zeros_like(S)
for i in range(S.shape[0]):
    for j in range(S.shape[1]):
        rate_landscape[i,j] = menten_hill_kinetics(S[i,j], k_max, K_m, N[i,j])

rate_landscape_gradient = np.gradient(rate_landscape)


# Plot 1
fig = figure(figsize=(15, 5))

# 1. Impact of cooperativity on reaction rate
ax1 = fig.add_subplot(1, 2, 1)

S_fine = np.linspace(0, 3.5, 300)
n_transition = [1.0, 2.0, 4.0]
colors = ['blue', 'lightsalmon', 'red']

for n, color in zip(n_transition, colors):
    rate = menten_hill_kinetics(S_fine, k_max, K_m, n)
    ax1.plot(S_fine, rate, color=color, linewidth=3, label=f'n={n}')

# Annotations
ax1.fill_between(S_fine, 0, 1, where=(S_fine<1), alpha=0.2, color='blue', label='S < K_m')
ax1.fill_between(S_fine, 0, 1, where=(S_fine>=1), alpha=0.2, color='red', label='S ≥ K_m')
ax1.axvline(x=1, color='black', linestyle='--', alpha=0.5, label='S=K_m=1')

ax1.set_xlabel('Substrate Concentration (S)')
ax1.set_ylabel('Reaction Rate')
ax1.set_title('Rate for differnt n')
ax1.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
ax1.grid(True, alpha=0.3)


# 2. Impact of cooperativity on derivative of reaction rate
ax2 = fig.add_subplot(1, 2, 2)

S_analysis = np.linspace(0.05, 3.5, 200)

for ind, n in enumerate([1.0, 2.0, 4.0]):
    rate = menten_hill_kinetics(S_analysis, k_max, K_m, n)
    derivative = np.gradient(rate, S_analysis)
    ax2.plot(S_analysis, derivative, label=f'n={n}', color=colors[ind], linewidth=3)

ax2.axvline(x=1, color='black', linestyle='--', alpha=0.7, label='S=K_m=1')
ax2.set_xlabel('Substrate Concentration (S)')
ax2.set_ylabel('Change of Rate')
ax2.set_title('Gradient for differnt n')
ax2.legend()
ax2.grid(True, alpha=0.3)

fig.tight_layout()


# Print analysis
print("Enzyme Kinetics without and with cooperativity")
print("=" * 50)
print(f"Parameters: k_max = {k_max}, K_m = {K_m}")
print("\nKey observations:")
print("1. n = 1: Classic hyperbolic Michaelis-Menten kinetics")
print("2. n > 1: Positive cooperativity - sigmoidal curves")
print("3. n >> 1: Switch-like behavior - very steep transition")
print(f"5. Half-maximal rate at S = K_m = {K_m}")
print("\n Change from hyperbolic (blue) to switch-like shape (orange, red)")
print(' ')
```

Note: the rate function (left) will always end at rate=1, independent of the Hill coefficient. Similarly the derivative (right) will always end at zero. 

## Two parameter view of the function

An extension of the above code lets us go from a single line which is evaluated at different values of n to a two-dimensional plane: we can evaluate the function for a range of S and a range of n simultaneously and plot the two-dimenstional result as a heat map.

```python
# Plot 2
fig = figure(figsize=(15, 5))

# 3. Rate Landscape (S vs n)
ax3 = fig.add_subplot(1, 2, 1)

im3 = ax3.imshow(rate_landscape, extent=[S_values.min(), S_values.max(), 
                                       n_values.min(), n_values.max()],
               origin='lower', aspect='auto', cmap='viridis')
ax3.set_xlabel('Substrate Concentration (S)')
ax3.set_ylabel('Hill Coefficient (n)')
ax3.set_title(f'Rate Landscape\n(k_max={k_max}, K_m={K_m})')

fig.colorbar(im3, ax=ax3, label='Rate', shrink=0.5)

contour_levels = [0.1, 0.3, 0.5, 0.7, 0.9]
contours = ax3.contour(S, N, rate_landscape, levels=contour_levels, 
                       colors='white', alpha=0.7, linewidths=1)
ax3.clabel(contours, inline=True, fontsize=8, fmt='%.1f')


# 4. Derivative of Rate Landscape
ax4 = fig.add_subplot(1, 2, 2)

im4 = ax4.imshow(rate_landscape_gradient[1], extent=[S_values.min(), S_values.max(), 
                                       n_values.min(), n_values.max()], vmax=0.03,
               origin='lower', aspect='auto', cmap='viridis')
ax4.set_xlabel('Substrate Concentration (S)')
ax4.set_ylabel('Hill Coefficient (n)')
ax4.set_title(f'Gradient Landscape\n(k_max={k_max}, K_m={K_m})')
fig.colorbar(im4, ax=ax4, label='Gradient', shrink=0.5)

# Add contour lines
contour_levels_gradient = [0.005, 0.01, 0.02, 0.03]
contours_grad = ax4.contour(S, N, rate_landscape_gradient[1], levels=contour_levels_gradient, 
                       colors='white', alpha=0.7, linewidths=1)
ax4.clabel(contours_grad, inline=True, fontsize=8, fmt='%.3f');
```

# Interactive Slider widget

```python
import plotly.graph_objects as go
```

```python
from plotly.subplots import make_subplots

S_values = np.linspace(0.001, 3.5, 300)
S_left   = np.linspace(0, 1,   100)
S_right  = np.linspace(1, 3.5, 150)

n_animation = np.linspace(0.5, 8.0, 16)

# ── subplots ──────────────────────────────────────────────────────────────────
fig_animation = make_subplots(
    rows=1, cols=2,
    subplot_titles=('Rate for varying n', 'Gradient for varying n'),
    horizontal_spacing=0.14
)

# ── trace 0: blue fill S < K_m (left subplot) ─────────────────────────────────
fig_animation.add_trace(go.Scatter(
    x=np.concatenate([S_left, S_left[::-1]]),
    y=np.concatenate([np.ones(100), np.zeros(100)]),
    fill='toself',
    fillcolor='rgba(0,0,255,0.2)',
    line=dict(color='rgba(0,0,0,0)'),
    name='S < K_m',
    hoverinfo='skip'
), row=1, col=1)

# ── trace 1: red fill S >= K_m (left subplot) ──────────────────────────────────
fig_animation.add_trace(go.Scatter(
    x=np.concatenate([S_right, S_right[::-1]]),
    y=np.concatenate([np.ones(150), np.zeros(150)]),
    fill='toself',
    fillcolor='rgba(255,0,0,0.2)',
    line=dict(color='rgba(0,0,0,0)'),
    name='S \u2265 K_m',
    hoverinfo='skip'
), row=1, col=1)

# ── trace 2: animated Hill function (left subplot) ───────────────────────────
n_init    = n_animation[0]
rate_init = menten_hill_kinetics(S_values, k_max, K_m, n_init)
deriv_init = np.gradient(rate_init, S_values)

fig_animation.add_trace(go.Scatter(
    x=S_values, y=rate_init,
    mode='lines',
    line=dict(color='red', width=3),
    name='Rate'
), row=1, col=1)

# ── trace 3: animated derivative (right subplot) ──────────────────────────────
fig_animation.add_trace(go.Scatter(
    x=S_values, y=deriv_init,
    mode='lines',
    line=dict(color='red', width=3),
    name='d(Rate)/dS'
), row=1, col=2)

# ── frames: update traces 2 & 3 together ─────────────────────────────────────
frames = []
for n_val in n_animation:
    rate  = menten_hill_kinetics(S_values, k_max, K_m, n_val)
    deriv = np.gradient(rate, S_values)
    frames.append(go.Frame(
        data=[
            go.Scatter(x=S_values, y=rate,  mode='lines', line=dict(color='red', width=3)),
            go.Scatter(x=S_values, y=deriv, mode='lines', line=dict(color='red', width=3))
        ],
        traces=[2, 3],
        name=f"n={n_val:.2f}"
    ))

fig_animation.frames = frames

# ── layout ────────────────────────────────────────────────────────────────────
fig_animation.update_layout(
    title=dict(
        text="Hill Kinetics: Effect of Cooperativity Parameter n"
             "<br><sub>Drag the slider to change Hill coefficient n</sub>",
        x=0.5
    ),
    width=1050,
    height=540,
    plot_bgcolor='white',
    legend=dict(x=1.02, y=1, xanchor='left'),
    shapes=[
        dict(type='line', x0=1, x1=1, y0=0, y1=1,
             xref='x',  yref='y domain',
             line=dict(color='black', dash='dash', width=1.5)),
        dict(type='line', x0=1, x1=1, y0=0, y1=1,
             xref='x2', yref='y2 domain',
             line=dict(color='black', dash='dash', width=1.5)),
    ],
    updatemenus=[dict(
        type='buttons',
        showactive=False,
        y=0,
        x=0.05,
        xanchor='right',
        buttons=[dict(
            label='\u25b6 Play',
            method='animate',
            args=[None, {'frame': {'duration': 100, 'redraw': True}, 'fromcurrent': True}]
        )]
    )],
    sliders=[dict(
        currentvalue=dict(prefix='n = ', font=dict(size=14)),
        pad=dict(t=50),
        steps=[dict(
            method='animate',
            args=[[f"n={n_val:.2f}"],
                  {'mode': 'immediate',
                   'frame': {'duration': 100, 'redraw': True},
                   'transition': {'duration': 0}}],
            label=f"{n_val:.2f}"
        ) for n_val in n_animation]
    )]
)

# axes styling to match static plot
grid_style = dict(gridcolor='rgba(0,0,0,0.15)', showgrid=True, gridwidth=1)

fig_animation.update_xaxes(title_text='Substrate Concentration (S)',
                            range=[-0.1, 3.6], **grid_style, row=1, col=1)
fig_animation.update_xaxes(title_text='Substrate Concentration (S)',
                            range=[-0.1, 3.6], **grid_style, row=1, col=2)
fig_animation.update_yaxes(title_text='Reaction Rate',
                            range=[-0.05, 1.15], **grid_style, row=1, col=1)
fig_animation.update_yaxes(title_text='Change of Rate',
                            range=[0, 2.2], **grid_style, row=1, col=2)

fig_animation.show()
```
