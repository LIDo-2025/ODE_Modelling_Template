# Symbolic Analysis of the Toggle Switch: Finding and Testing Steady States

This add-on tutorial shows how to symbolically analyze the genetic toggle switch model to find its steady states and determine their stability. We'll use Python's SymPy library to perform the algebra, making the mathematical concepts accessible through code and visualization.

## 1. Understanding Steady States Mathematically

For the toggle switch model:
```
du/dt = α₁/(1 + vⁿ) - βu
dv/dt = α₂/(1 + uⁿ) - γv
```

A **steady state** occurs when concentrations don't change: `du/dt = 0` and `dv/dt = 0`. This gives us two equations:
```
α₁/(1 + vⁿ) = βu
α₂/(1 + uⁿ) = γv
```

Solving these equations simultaneously gives us the fixed points of the system.

## 2. Complete Symbolic Analysis Code

Here's the Python code that performs the symbolic analysis:

````python

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from IPython.display import display, Math

# 1. SET UP SYMBOLIC VARIABLES AND EQUATIONS
print("=== SYMBOLIC ANALYSIS OF TOGGLE SWITCH ===")
print("\n1. Defining symbolic variables and equations...")

# Create symbolic variables
u, v = sp.symbols('u v', positive=True)  # concentrations must be positive
alpha1, alpha2, beta, gamma, n = sp.symbols('alpha1 alpha2 beta gamma n', positive=True)

# Define the ODEs
du_dt = alpha1/(1 + v**n) - beta*u
dv_dt = alpha2/(1 + u**n) - gamma*v

print(f"ODEs:")
display(Math(fr"\frac{{du}}{{dt}} = {sp.latex(du_dt)}"))
display(Math(fr"\frac{{dv}}{{dt}} = {sp.latex(dv_dt)}"))

# 2. FIND STEADY STATES BY SOLVING du/dt = 0 AND dv/dt = 0
print("\n2. Finding steady states (solve du/dt = 0, dv/dt = 0)...")

# Solve for steady states symbolically
steady_states = sp.solve([du_dt, dv_dt], [u, v], dict=True)
print(f"\nFound {len(steady_states)} steady state(s):")

for i, sol in enumerate(steady_states):
    print(f"\nSteady State {i+1}:")
    display(Math(fr"u^* = {sp.latex(sol[u])}"))
    display(Math(fr"v^* = {sp.latex(sol[v])}"))

# 3. EVALUATE FOR SPECIFIC PARAMETERS (NUMERIC EXAMPLE)
print("\n3. Evaluating for specific parameter values...")

# Use the same parameters as our simulation
param_values = {
    alpha1: 10.0,
    alpha2: 10.0,
    beta: 1.0,
    gamma: 1.0,
    n: 2.0
}

print("Parameter values:")
for param, value in param_values.items():
    print(f"  {param} = {value}")

# Calculate numeric steady states
numeric_states = []
for sol in steady_states:
    u_val = float(sol[u].subs(param_values))
    v_val = float(sol[v].subs(param_values))
    numeric_states.append((u_val, v_val))
    print(f"\nNumeric steady state: u = {u_val:.4f}, v = {v_val:.4f}")

# 4. ANALYZE STABILITY: CALCULATE THE JACOBIAN MATRIX
print("\n4. Analyzing stability using the Jacobian matrix...")

# Create the Jacobian matrix (matrix of partial derivatives)
J = sp.Matrix([[sp.diff(du_dt, u), sp.diff(du_dt, v)],
               [sp.diff(dv_dt, u), sp.diff(dv_dt, v)]])

print("Jacobian matrix J = ")
display(Math(fr"J = \begin{{bmatrix}} \frac{{\partial f_1}}{{\partial u}} & \frac{{\partial f_1}}{{\partial v}} \\ \frac{{\partial f_2}}{{\partial u}} & \frac{{\partial f_2}}{{\partial v}} \end{{bmatrix}} = {sp.latex(J)}"))

# 5. EVALUATE JACOBIAN AT EACH STEADY STATE
print("\n5. Evaluating Jacobian at each steady state...")

for i, (u_val, v_val) in enumerate(numeric_states):
    print(f"\n--- Steady State {i+1}: (u = {u_val:.4f}, v = {v_val:.4f}) ---")
    
    # Substitute the steady state values and parameters
    J_at_state = J.subs({u: u_val, v: v_val, **param_values})
    
    # Convert to numeric matrix
    J_numeric = np.array(J_at_state).astype(float)
    print(f"Jacobian at this state:")
    print(J_numeric)
    
    # Calculate eigenvalues (determine stability)
    eigenvalues = np.linalg.eigvals(J_numeric)
    print(f"Eigenvalues: {eigenvalues[0]:.4f}, {eigenvalues[1]:.4f}")
    
    # Determine stability based on eigenvalues
    # If all eigenvalues have negative real parts: stable
    # If any eigenvalue has positive real part: unstable
    real_parts = eigenvalues.real
    if all(real_parts < 0):
        stability = "STABLE (all eigenvalues negative)"
    elif any(real_parts > 0):
        stability = "UNSTABLE (at least one positive eigenvalue)"
    else:
        stability = "MARGINALLY STABLE (on imaginary axis)"
    
    print(f"Stability: {stability}")
    
    # Also calculate the determinant and trace
    det = np.linalg.det(J_numeric)
    trace = np.trace(J_numeric)
    print(f"Determinant: {det:.4f}, Trace: {trace:.4f}")
    
    # For 2D systems: stable if det > 0 and trace < 0
    if det > 0 and trace < 0:
        print("2D stability condition satisfied: det > 0 and trace < 0")

# 6. VISUALIZE STEADY STATES AND STABILITY
print("\n6. Visualizing steady states and their stability...")

# Create a grid for visualization
u_vals = np.linspace(0, 12, 400)
v_vals = np.linspace(0, 12, 400)

# Create the figure
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# LEFT PLOT: NULLCLINES WITH STEADY STATES
# Nullcline du/dt = 0: u = α₁/(β(1+vⁿ))
v_for_nullcline = np.linspace(0, 12, 400)
u_nullcline = param_values[alpha1]/(param_values[beta]*(1 + v_for_nullcline**param_values[n]))

# Nullcline dv/dt = 0: v = α₂/(γ(1+uⁿ))
u_for_nullcline = np.linspace(0, 12, 400)
v_nullcline = param_values[alpha2]/(param_values[gamma]*(1 + u_for_nullcline**param_values[n]))

axes[0].plot(u_nullcline, v_for_nullcline, 'b-', linewidth=2.5, label='du/dt = 0 nullcline')
axes[0].plot(u_for_nullcline, v_nullcline, 'r-', linewidth=2.5, label='dv/dt = 0 nullcline')

# Mark the steady states
for i, (u_val, v_val) in enumerate(numeric_states):
    # Color code by stability
    J_at_state = J.subs({u: u_val, v: v_val, **param_values})
    J_numeric = np.array(J_at_state).astype(float)
    eigenvalues = np.linalg.eigvals(J_numeric)
    
    if all(eigenvalues.real < 0):
        color = 'green'
        marker = 'o'
        label = f'Stable SS {i+1}'
        size = 120
    else:
        color = 'red'
        marker = 'x'
        label = f'Unstable SS {i+1}'
        size = 100
    
    axes[0].scatter(u_val, v_val, color=color, marker=marker, s=size, 
                   edgecolors='black', linewidth=2, zorder=5, label=label)

axes[0].set_xlabel('u (Repressor 1 Concentration)', fontsize=12)
axes[0].set_ylabel('v (Repressor 2 Concentration)', fontsize=12)
axes[0].set_title('Nullclines and Steady States', fontsize=14)
axes[0].legend(fontsize=10, loc='upper right')
axes[0].grid(True, alpha=0.3)
axes[0].set_xlim(0, 12)
axes[0].set_ylim(0, 12)

# RIGHT PLOT: VECTOR FIELD WITH STEADY STATES
# Calculate vector field
U, V = np.meshgrid(np.linspace(0, 12, 15), np.linspace(0, 12, 15))
DU = np.zeros_like(U)
DV = np.zeros_like(V)

for i in range(U.shape[0]):
    for j in range(U.shape[1]):
        u_pt, v_pt = U[i, j], V[i, j]
        DU[i, j] = param_values[alpha1]/(1 + v_pt**param_values[n]) - param_values[beta]*u_pt
        DV[i, j] = param_values[alpha2]/(1 + u_pt**param_values[n]) - param_values[gamma]*v_pt

# Normalize arrows for better visualization
magnitude = np.sqrt(DU**2 + DV**2)
DU_norm = DU / (magnitude + 1e-10)  # Add small constant to avoid division by zero
DV_norm = DV / (magnitude + 1e-10)

axes[1].quiver(U, V, DU_norm, DV_norm, magnitude, cmap='viridis', 
              scale=30, width=0.004, alpha=0.7)

# Mark steady states again
for i, (u_val, v_val) in enumerate(numeric_states):
    J_at_state = J.subs({u: u_val, v: v_val, **param_values})
    J_numeric = np.array(J_at_state).astype(float)
    eigenvalues = np.linalg.eigvals(J_numeric)
    
    if all(eigenvalues.real < 0):
        color = 'green'
        marker = 'o'
        label = 'Stable' if i == 0 else ''
    else:
        color = 'red'
        marker = 'x'
        label = 'Unstable' if i == 0 else ''
    
    axes[1].scatter(u_val, v_val, color=color, marker=marker, s=100, 
                   edgecolors='black', linewidth=2, zorder=5, label=label)

axes[1].set_xlabel('u (Repressor 1 Concentration)', fontsize=12)
axes[1].set_ylabel('v (Repressor 2 Concentration)', fontsize=12)
axes[1].set_title('Vector Field and Steady State Stability', fontsize=14)
axes[1].legend(fontsize=10, loc='upper right')
axes[1].grid(True, alpha=0.3)
axes[1].set_xlim(0, 12)
axes[1].set_ylim(0, 12)

plt.tight_layout()
plt.show()

# 7. ADDITIONAL INSIGHT: BIFURCATION ANALYSIS
print("\n7. Bonus: How bistability depends on parameters (bifurcation analysis)...")
print("Let's see what happens when we change the cooperativity parameter n:")

# Test different n values
n_values = [1.0, 1.5, 2.0, 3.0]
print("\nNumber of steady states for different n values:")

for n_val in n_values:
    temp_params = param_values.copy()
    temp_params[n] = n_val
    
    # Find steady states for this n
    states_for_n = []
    for sol in steady_states:
        try:
            u_val = float(sol[u].subs({alpha1: temp_params[alpha1], 
                                      alpha2: temp_params[alpha2],
                                      beta: temp_params[beta],
                                      gamma: temp_params[gamma],
                                      n: n_val}))
            v_val = float(sol[v].subs({alpha1: temp_params[alpha1],
                                      alpha2: temp_params[alpha2],
                                      beta: temp_params[beta],
                                      gamma: temp_params[gamma],
                                      n: n_val}))
            
            # Check if it's a valid positive steady state
            if u_val > 0 and v_val > 0:
                states_for_n.append((u_val, v_val))
        except:
            pass
    
    print(f"  n = {n_val}: {len(states_for_n)} steady state(s)")
    for u_val, v_val in states_for_n:
        print(f"    u = {u_val:.3f}, v = {v_val:.3f}")

print("\nNote: Bistability (2 stable states) requires n > 1 (cooperativity).")
````

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve
from scipy.integrate import solve_ivp

print("=== SYMBOLIC-NUMERIC ANALYSIS OF TOGGLE SWITCH ===")
print("We'll set up equations symbolically but solve them numerically.")
print("This is more realistic for complex biological models.\n")

# 1. SYMBOLIC SETUP (for understanding)
print("1. Symbolic setup of the equations...")

# Create symbolic variables
u, v = sp.symbols('u v', positive=True)
alpha1, alpha2, beta, gamma, n = sp.symbols('alpha1 alpha2 beta gamma n', positive=True)

# Steady state equations: du/dt = 0 and dv/dt = 0
eq1 = sp.Eq(alpha1/(1 + v**n), beta*u)
eq2 = sp.Eq(alpha2/(1 + u**n), gamma*v)

print("\nSteady state equations:")
print(f"1. {eq1}")
print(f"2. {eq2}")

# Solve equation 1 for u in terms of v
u_expr = sp.solve(eq1, u)[0]
print(f"\nFrom equation 1, we can express u in terms of v:")
print(f"u = {u_expr}")

# Solve equation 2 for v in terms of u  
v_expr = sp.solve(eq2, v)[0]
print(f"\nFrom equation 2, we can express v in terms of u:")
print(f"v = {v_expr}")

# 2. NUMERICAL SOLUTION FOR SPECIFIC PARAMETERS
print("\n" + "="*60)
print("2. Numerical solution for specific parameters...")

# Parameter values (same as before)
params = {
    'alpha1': 10.0,
    'alpha2': 10.0, 
    'beta': 1.0,
    'gamma': 1.0,
    'n': 2.0
}

print(f"Parameters: {params}")

# Define the equations numerically for fsolve
def steady_state_eqs(vars, *args):
    """Numerical equations for steady states"""
    u_val, v_val = vars
    alpha1, alpha2, beta, gamma, n = args
    
    # Equations: f1 = du/dt, f2 = dv/dt
    f1 = alpha1/(1 + v_val**n) - beta*u_val
    f2 = alpha2/(1 + u_val**n) - gamma*v_val
    
    return [f1, f2]

# Find steady states using different initial guesses
print("\nSearching for steady states with different initial guesses...")

# Initial guesses that should find different steady states
initial_guesses = [
    [0.1, 10],    # Should find State 2 (low u, high v)
    [10, 0.1],    # Should find State 1 (high u, low v)  
    [3, 3]        # Should find the unstable state (if it exists)
]

steady_states = []
for i, guess in enumerate(initial_guesses):
    try:
        # Use fsolve to find roots
        args = (params['alpha1'], params['alpha2'], params['beta'], 
                params['gamma'], params['n'])
        
        solution = fsolve(steady_state_eqs, guess, args=args, full_output=True)
        u_sol, v_sol = solution[0]
        
        # Check if solution is valid (positive and not duplicate)
        valid = True
        if u_sol < 0 or v_sol < 0:
            valid = False
        
        # Check if it's essentially a duplicate
        for existing in steady_states:
            if np.abs(u_sol - existing[0]) < 0.1 and np.abs(v_sol - existing[1]) < 0.1:
                valid = False
                break
        
        if valid:
            steady_states.append((u_sol, v_sol))
            print(f"  Guess {i+1}: [{guess[0]}, {guess[1]}] -> Found: u={u_sol:.4f}, v={v_sol:.4f}")
        else:
            print(f"  Guess {i+1}: [{guess[0]}, {guess[1]}] -> Invalid or duplicate")
            
    except Exception as e:
        print(f"  Guess {i+1}: Failed - {e}")

print(f"\nFound {len(steady_states)} distinct steady state(s):")
for i, (u_val, v_val) in enumerate(steady_states):
    print(f"  State {i+1}: u* = {u_val:.4f}, v* = {v_val:.4f}")

# 3. STABILITY ANALYSIS (NUMERICAL JACOBIAN)
print("\n" + "="*60)
print("3. Numerical stability analysis...")

def jacobian_at_point(u_val, v_val, params):
    """Calculate Jacobian matrix numerically at a point"""
    n = params['n']
    
    # Partial derivatives (calculated manually from our equations)
    # du/dt = α₁/(1 + vⁿ) - βu
    # ∂(du/dt)/∂u = -β
    # ∂(du/dt)/∂v = -α₁ * n * v^(n-1) / (1 + vⁿ)²
    
    # dv/dt = α₂/(1 + uⁿ) - γv  
    # ∂(dv/dt)/∂u = -α₂ * n * u^(n-1) / (1 + uⁿ)²
    # ∂(dv/dt)/∂v = -γ
    
    J = np.zeros((2, 2))
    
    # Top row: derivatives of du/dt
    J[0, 0] = -params['beta']  # ∂(du/dt)/∂u
    J[0, 1] = -params['alpha1'] * n * v_val**(n-1) / (1 + v_val**n)**2  # ∂(du/dt)/∂v
    
    # Bottom row: derivatives of dv/dt  
    J[1, 0] = -params['alpha2'] * n * u_val**(n-1) / (1 + u_val**n)**2  # ∂(dv/dt)/∂u
    J[1, 1] = -params['gamma']  # ∂(dv/dt)/∂v
    
    return J

print("\nStability analysis for each steady state:")
for i, (u_val, v_val) in enumerate(steady_states):
    print(f"\n--- State {i+1}: u={u_val:.4f}, v={v_val:.4f} ---")
    
    # Calculate Jacobian
    J = jacobian_at_point(u_val, v_val, params)
    print(f"Jacobian matrix:")
    print(f"  [{J[0,0]:.4f}, {J[0,1]:.4f}]")
    print(f"  [{J[1,0]:.4f}, {J[1,1]:.4f}]")
    
    # Calculate eigenvalues
    eigenvalues = np.linalg.eigvals(J)
    print(f"Eigenvalues: {eigenvalues[0]:.4f}, {eigenvalues[1]:.4f}")
    
    # Determine stability
    real_parts = eigenvalues.real
    if all(real_parts < 0):
        stability = "STABLE (sink)"
        stability_symbol = "●"
    elif all(real_parts > 0):
        stability = "UNSTABLE (source)"
        stability_symbol = "⨉"
    elif real_parts[0] * real_parts[1] < 0:
        stability = "UNSTABLE (saddle point)"
        stability_symbol = "⨉"
    else:
        stability = "MARGINALLY STABLE"
        stability_symbol = "○"
    
    print(f"Stability: {stability} {stability_symbol}")

# 4. VISUALIZATION: NULLCLINES AND STEADY STATES
print("\n" + "="*60)
print("4. Visualizing nullclines and steady states...")

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Generate nullcline curves
v_range = np.linspace(0, 12, 400)
u_nullcline = params['alpha1']/(params['beta'] * (1 + v_range**params['n']))

u_range = np.linspace(0, 12, 400)  
v_nullcline = params['alpha2']/(params['gamma'] * (1 + u_range**params['n']))

# Plot nullclines
axes[0].plot(u_nullcline, v_range, 'b-', linewidth=2, label='du/dt = 0')
axes[0].plot(u_range, v_nullcline, 'r-', linewidth=2, label='dv/dt = 0')

# Plot steady states with stability markers
for i, (u_val, v_val) in enumerate(steady_states):
    # Calculate stability for coloring
    J = jacobian_at_point(u_val, v_val, params)
    eigenvalues = np.linalg.eigvals(J)
    real_parts = eigenvalues.real
    
    if all(real_parts < 0):
        color = 'green'
        marker = 'o'
        label = f'Stable State {i+1}'
        size = 100
    else:
        color = 'red'
        marker = 'd'
        label = f'Unstable State {i+1}'
        size = 80
    
    axes[0].scatter(u_val, v_val, color=color, marker=marker, s=size,
                   edgecolors='black', linewidth=2, zorder=5, label=label)

axes[0].set_xlabel('u (Repressor 1)', fontsize=12)
axes[0].set_ylabel('v (Repressor 2)', fontsize=12)
axes[0].set_title('Nullclines and Steady States', fontsize=14)
axes[0].legend(fontsize=10, loc='upper right')
axes[0].grid(True, alpha=0.3)
axes[0].set_xlim(0, 12)
axes[0].set_ylim(0, 12)

# 5. VISUALIZATION 2: BASINS OF ATTRACTION
print("\n5. Visualizing basins of attraction...")

# Create a grid of initial conditions
u_grid, v_grid = np.meshgrid(np.linspace(0.1, 11.9, 30), 
                             np.linspace(0.1, 11.9, 30))

# Determine which steady state each IC goes to
basin_map = np.zeros_like(u_grid, dtype=int)

for i in range(u_grid.shape[0]):
    for j in range(u_grid.shape[1]):
        # Integrate from this initial condition
        ic = [u_grid[i, j], v_grid[i, j]]
        sol = solve_ivp(
            lambda t, y: [
                params['alpha1']/(1 + y[1]**params['n']) - params['beta']*y[0],
                params['alpha2']/(1 + y[0]**params['n']) - params['gamma']*y[1]
            ],
            [0, 100], ic, max_step=0.1
        )
        
        # Check which steady state it's closest to at the end
        final_u, final_v = sol.y[0, -1], sol.y[1, -1]
        
        # Find closest steady state
        distances = []
        for k, (u_ss, v_ss) in enumerate(steady_states):
            dist = np.sqrt((final_u - u_ss)**2 + (final_v - v_ss)**2)
            distances.append((dist, k))
        
        # Use the closest one
        distances.sort()
        basin_map[i, j] = distances[0][1] + 1  # 1, 2, or 3

# Plot basins of attraction
basin_colors = ['lightblue', 'lightgreen', 'yellow']
for basin_num in range(1, len(steady_states)+1):
    mask = basin_map == basin_num
    axes[1].scatter(u_grid[mask], v_grid[mask], 
                   color=basin_colors[basin_num-1], 
                   s=30, alpha=0.7, 
                   label=f'Basin for State {basin_num}')

# Overlay steady states
for i, (u_val, v_val) in enumerate(steady_states):
    J = jacobian_at_point(u_val, v_val, params)
    eigenvalues = np.linalg.eigvals(J)
    real_parts = eigenvalues.real
    
    if all(real_parts < 0):
        color = 'darkgreen'
        marker = 'o'
    else:
        color = 'darkred'
        marker = 'd'
    
    axes[1].scatter(u_val, v_val, color=color, marker=marker, s=120,
                   edgecolors='black', linewidth=2, zorder=5)

axes[1].set_xlabel('u (Repressor 1)', fontsize=12)
axes[1].set_ylabel('v (Repressor 2)', fontsize=12)
axes[1].set_title('Basins of Attraction', fontsize=14)
axes[1].legend(fontsize=10, loc='upper right')
axes[1].grid(True, alpha=0.3)
axes[1].set_xlim(0, 12)
axes[1].set_ylim(0, 12)

plt.tight_layout()
plt.show()

# 6. INTERACTIVE EXPLORATION
print("\n" + "="*60)
print("6. Interactive exploration: How cooperativity affects bistability")

# Test different n values
print("\nNumber of stable states for different cooperativity (n) values:")

n_values = [1.0, 1.5, 2.0, 2.5, 3.0]
stable_counts = []

for n_val in n_values:
    # Update parameters
    test_params = params.copy()
    test_params['n'] = n_val
    
    # Find steady states for this n
    test_states = []
    test_guesses = [[0.1, 10], [10, 0.1], [3, 3], [1, 1], [5, 5]]
    
    for guess in test_guesses:
        args = (test_params['alpha1'], test_params['alpha2'], test_params['beta'],
                test_params['gamma'], test_params['n'])
        
        try:
            sol = fsolve(steady_state_eqs, guess, args=args)
            u_sol, v_sol = sol
            
            if u_sol > 0 and v_sol > 0:
                # Check if new
                is_new = True
                for existing in test_states:
                    if np.abs(u_sol - existing[0]) < 0.2 and np.abs(v_sol - existing[1]) < 0.2:
                        is_new = False
                        break
                
                if is_new:
                    test_states.append((u_sol, v_sol))
        except:
            pass
    
    # Count stable ones
    stable = 0
    for u_sol, v_sol in test_states:
        J = jacobian_at_point(u_sol, v_sol, test_params)
        eigenvalues = np.linalg.eigvals(J)
        if all(eigenvalues.real < 0):
            stable += 1
    
    stable_counts.append(stable)
    print(f"  n = {n_val}: {len(test_states)} steady states, {stable} stable")

# Plot the results
fig2, ax2 = plt.subplots(figsize=(8, 5))
ax2.plot(n_values, stable_counts, 'bo-', linewidth=2, markersize=8)
ax2.axhline(y=2, color='r', linestyle='--', alpha=0.5, label='Bistability threshold')
ax2.set_xlabel('Cooperativity (n)', fontsize=12)
ax2.set_ylabel('Number of Stable States', fontsize=12)
ax2.set_title('Bistability Emerges with Cooperativity', fontsize=14)
ax2.grid(True, alpha=0.3)
ax2.legend(fontsize=10)
ax2.set_xticks(n_values)

plt.tight_layout()
plt.show()

print("\n" + "="*60)
print("KEY INSIGHTS:")
print("1. The toggle switch has MULTIPLE steady states when n > 1")
print("2. For our symmetric parameters (α₁=α₂, β=γ):")
print("   - 2 stable states (green circles): the 'memory' states")
print("   - 1 unstable state (red X): the switching threshold")
print("3. Bistability REQUIRES cooperativity (n > 1)")
print("4. The basins of attraction show which initial conditions lead to which final state")
````

````python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve
from scipy.integrate import solve_ivp

print("=== SYMBOLIC-NUMERIC ANALYSIS OF GENETIC TOGGLE SWITCH ===")
print("Combining symbolic understanding with numerical computation")
print("=" * 60)

# 1. SYMBOLIC SETUP FOR UNDERSTANDING
print("\n1. SYMBOLIC SETUP: Understanding the equations")

# Create symbolic variables
u, v = sp.symbols('u v', positive=True)
alpha1, alpha2, beta, gamma, n = sp.symbols('alpha1 alpha2 beta gamma n', positive=True)

# Steady state equations
print("\nAt steady state, concentrations don't change:")
print("  du/dt = 0  and  dv/dt = 0")
print()
print("Which gives us:")
print(f"  α₁/(1 + vⁿ) = β·u")
print(f"  α₂/(1 + uⁿ) = γ·v")

# 2. NUMERICAL ANALYSIS WITH BIOLOGICAL PARAMETERS
print("\n" + "=" * 60)
print("2. NUMERICAL ANALYSIS: Finding actual steady states")

# Parameter values (biological interpretation)
params = {
    'alpha1': 10.0,  # Max production rate of repressor 1
    'alpha2': 10.0,  # Max production rate of repressor 2  
    'beta': 1.0,     # Degradation rate of repressor 1
    'gamma': 1.0,    # Degradation rate of repressor 2
    'n': 2.0         # Cooperativity (Hill coefficient)
}

print(f"\nBiological parameters:")
print(f"  α₁ = {params['alpha1']}: Max production rate of repressor 1")
print(f"  α₂ = {params['alpha2']}: Max production rate of repressor 2")
print(f"  β = {params['beta']}, γ = {params['gamma']}: Degradation rates")
print(f"  n = {params['n']}: Cooperativity (how many repressors needed to turn off gene)")

# Numerical function for finding steady states
def find_steady_states(params, initial_guesses=None):
    """Find all steady states of the toggle switch"""
    if initial_guesses is None:
        initial_guesses = [
            [0.1, 10],   # Should find State 2 (low u, high v)
            [10, 0.1],   # Should find State 1 (high u, low v)
            [3, 3],      # Should find unstable state
            [1, 1],      # Additional check
            [5, 5]       # Additional check
        ]
    
    def equations(vars):
        u_val, v_val = vars
        return [
            params['alpha1']/(1 + v_val**params['n']) - params['beta']*u_val,
            params['alpha2']/(1 + u_val**params['n']) - params['gamma']*v_val
        ]
    
    steady_states = []
    
    for guess in initial_guesses:
        try:
            solution = fsolve(equations, guess, full_output=True)
            u_sol, v_sol = solution[0]
            
            # Check validity
            if u_sol > 0 and v_sol > 0:  # Must be positive
                # Check if it's essentially new (not a duplicate)
                is_new = True
                for existing in steady_states:
                    dist = np.sqrt((u_sol - existing[0])**2 + (v_sol - existing[1])**2)
                    if dist < 0.1:  # Too close to existing point
                        is_new = False
                        break
                
                if is_new:
                    steady_states.append((u_sol, v_sol))
        except:
            pass
    
    return steady_states

# Find all steady states
steady_states = find_steady_states(params)
print(f"\nFound {len(steady_states)} steady state(s):")

for i, (u_val, v_val) in enumerate(steady_states):
    print(f"  State {i+1}: u* = {u_val:.4f}, v* = {v_val:.4f}")

# 3. STABILITY ANALYSIS: VALLEYS vs HILLTOPS
print("\n" + "=" * 60)
print("3. STABILITY ANALYSIS: Biological stability of each state")

def analyze_stability(u_val, v_val, params):
    """Analyze stability of a steady state"""
    n = params['n']
    
    # Calculate Jacobian matrix (local linear approximation)
    J = np.zeros((2, 2))
    
    # Top row: derivatives of du/dt
    J[0, 0] = -params['beta']  # ∂(du/dt)/∂u
    J[0, 1] = -params['alpha1'] * n * v_val**(n-1) / (1 + v_val**n)**2
    
    # Bottom row: derivatives of dv/dt
    J[1, 0] = -params['alpha2'] * n * u_val**(n-1) / (1 + u_val**n)**2
    J[1, 1] = -params['gamma']
    
    # Calculate eigenvalues
    eigenvalues = np.linalg.eigvals(J)
    
    # Biological interpretation
    if all(eigenvalues.real < -0.01):  # All negative
        stability = "STABLE"
        interpretation = "Like a valley: small perturbations return to this state"
    elif all(eigenvalues.real > 0.01):  # All positive
        stability = "UNSTABLE (Source)"
        interpretation = "Like a hilltop: any perturbation grows"
    elif eigenvalues.real[0] * eigenvalues.real[1] < -0.01:  # One positive, one negative
        stability = "UNSTABLE (Saddle)"
        interpretation = "Like a mountain pass: stable in one direction, unstable in another"
    else:
        stability = "MARGINALLY STABLE"
        interpretation = "On the edge of stability"
    
    return {
        'jacobian': J,
        'eigenvalues': eigenvalues,
        'stability': stability,
        'interpretation': interpretation,
        'determinant': np.linalg.det(J),
        'trace': np.trace(J)
    }

print("\nStability analysis for each state:")
for i, (u_val, v_val) in enumerate(steady_states):
    analysis = analyze_stability(u_val, v_val, params)
    
    print(f"\n--- State {i+1}: u={u_val:.3f}, v={v_val:.3f} ---")
    print(f"  Stability: {analysis['stability']}")
    print(f"  Eigenvalues: {analysis['eigenvalues'][0]:.3f}, {analysis['eigenvalues'][1]:.3f}")
    print(f"  Interpretation: {analysis['interpretation']}")

# 4. ENHANCED VISUALIZATION WITH PROPER MARKERS
print("\n" + "=" * 60)
print("4. VISUALIZATION: Seeing the biological dynamics")

# Create figure with subplots
fig = plt.figure(figsize=(16, 10))

# Plot 1: Nullclines and steady states
ax1 = plt.subplot(2, 3, 1)

# Generate nullclines
v_range = np.linspace(0.01, 12, 400)
u_nullcline = params['alpha1']/(params['beta'] * (1 + v_range**params['n']))

u_range = np.linspace(0.01, 12, 400)
v_nullcline = params['alpha2']/(params['gamma'] * (1 + u_range**params['n']))

# Plot nullclines
ax1.plot(u_nullcline, v_range, 'b-', linewidth=2.5, alpha=0.8, label='du/dt = 0 (u nullcline)')
ax1.plot(u_range, v_nullcline, 'r-', linewidth=2.5, alpha=0.8, label='dv/dt = 0 (v nullcline)')

# Plot steady states with FIXED markers (no edgecolor for 'x')
for i, (u_val, v_val) in enumerate(steady_states):
    analysis = analyze_stability(u_val, v_val, params)
    
    if analysis['stability'] == "STABLE":
        # Filled green circle for stable states
        ax1.scatter(u_val, v_val, color='green', marker='o', s=200, 
                   zorder=5, label=f'Stable State {i+1}' if i == 0 else "")
        # Add black edge manually for filled markers
        ax1.scatter(u_val, v_val, color='black', marker='o', s=210, 
                   facecolors='none', linewidth=2, zorder=4)
    else:
        # Red 'x' for unstable states (no edgecolor)
        ax1.scatter(u_val, v_val, color='red', marker='x', s=150, 
                   linewidths=3, zorder=5, 
                   label=f'Unstable State {i+1}' if i == len(steady_states)-1 else "")

ax1.set_xlabel('u (Repressor 1 Concentration)', fontsize=11)
ax1.set_ylabel('v (Repressor 2 Concentration)', fontsize=11)
ax1.set_title('Nullclines and Steady States', fontsize=13, fontweight='bold')
ax1.legend(fontsize=9, loc='upper right')
ax1.grid(True, alpha=0.2)
ax1.set_xlim(0, 12)
ax1.set_ylim(0, 12)

# Plot 2: Vector field
ax2 = plt.subplot(2, 3, 2)

# Create vector field
U, V = np.meshgrid(np.linspace(0.5, 11.5, 15), np.linspace(0.5, 11.5, 15))
DU = np.zeros_like(U)
DV = np.zeros_like(V)

for i in range(U.shape[0]):
    for j in range(U.shape[1]):
        u_pt, v_pt = U[i, j], V[i, j]
        DU[i, j] = params['alpha1']/(1 + v_pt**params['n']) - params['beta']*u_pt
        DV[i, j] = params['alpha2']/(1 + u_pt**params['n']) - params['gamma']*v_pt

# Normalize for better visualization
magnitude = np.sqrt(DU**2 + DV**2)
DU_norm = DU / (magnitude + 1e-10)
DV_norm = DV / (magnitude + 1e-10)

ax2.quiver(U, V, DU_norm, DV_norm, magnitude, cmap='viridis', 
          scale=40, width=0.004, alpha=0.7)

# Plot steady states on vector field
for i, (u_val, v_val) in enumerate(steady_states):
    analysis = analyze_stability(u_val, v_val, params)
    
    if analysis['stability'] == "STABLE":
        ax2.scatter(u_val, v_val, color='green', marker='o', s=120, zorder=5)
        ax2.scatter(u_val, v_val, color='black', marker='o', s=130, 
                   facecolors='none', linewidth=1.5, zorder=4)
    else:
        ax2.scatter(u_val, v_val, color='red', marker='x', s=100, 
                   linewidths=2.5, zorder=5)

ax2.set_xlabel('u (Repressor 1)', fontsize=11)
ax2.set_ylabel('v (Repressor 2)', fontsize=11)
ax2.set_title('Vector Field: Direction of Change', fontsize=13, fontweight='bold')
ax2.grid(True, alpha=0.2)
ax2.set_xlim(0, 12)
ax2.set_ylim(0, 12)

# Plot 3: Time courses from different starting points
ax3 = plt.subplot(2, 3, 3)

# Test different initial conditions
test_ics = [
    ([0.5, 11.5], 'Starting: low u, high v', '#1f77b4'),
    ([11.5, 0.5], 'Starting: high u, low v', '#ff7f0e'),
    ([3.0, 3.0], 'Starting near unstable point', '#2ca02c'),
    ([1.0, 6.0], 'Starting in middle', '#d62728')
]

for ic, label, color in test_ics:
    sol = solve_ivp(
        lambda t, y: [
            params['alpha1']/(1 + y[1]**params['n']) - params['beta']*y[0],
            params['alpha2']/(1 + y[0]**params['n']) - params['gamma']*y[1]
        ],
        [0, 50], ic, max_step=0.1, dense_output=True
    )
    
    ax3.plot(sol.t, sol.y[0], color=color, linestyle='-', linewidth=2, 
             alpha=0.8, label=f'{label} (u)')
    ax3.plot(sol.t, sol.y[1], color=color, linestyle='--', linewidth=2, 
             alpha=0.5, label=f'{label} (v)')

ax3.set_xlabel('Time', fontsize=11)
ax3.set_ylabel('Concentration', fontsize=11)
ax3.set_title('Time Courses from Different Starts', fontsize=13, fontweight='bold')
ax3.legend(fontsize=8, loc='upper right')
ax3.grid(True, alpha=0.2)

# Plot 4: Basins of attraction
ax4 = plt.subplot(2, 3, 4, projection='3d')

# Create a potential-like surface
u_grid, v_grid = np.meshgrid(np.linspace(0.1, 11.9, 50), 
                             np.linspace(0.1, 11.9, 50))

# Calculate a "potential" function (simplified)
potential = np.zeros_like(u_grid)
for i in range(u_grid.shape[0]):
    for j in range(u_grid.shape[1]):
        # Simplified potential for visualization
        u_val, v_val = u_grid[i, j], v_grid[i, j]
        # This isn't the true potential but gives the right shape
        pot1 = params['beta'] * u_val**2 / 2 - params['alpha1'] * np.log(1 + v_val**params['n'])
        pot2 = params['gamma'] * v_val**2 / 2 - params['alpha2'] * np.log(1 + u_val**params['n'])
        potential[i, j] = pot1 + pot2

# Plot the surface
surf = ax4.plot_surface(u_grid, v_grid, potential, cmap='viridis', 
                       alpha=0.7, edgecolor='none')

# Mark the steady states on the surface
for i, (u_val, v_val) in enumerate(steady_states):
    # Find the potential value at this point
    pot_val = np.interp(u_val, u_grid[0,:], potential[int(v_val/12*49), :])
    
    analysis = analyze_stability(u_val, v_val, params)
    
    if analysis['stability'] == "STABLE":
        ax4.scatter(u_val, v_val, pot_val, color='green', marker='o', 
                   s=100, edgecolor='black', linewidth=1.5, zorder=5)
    else:
        ax4.scatter(u_val, v_val, pot_val, color='red', marker='^', 
                   s=100, edgecolor='black', linewidth=1.5, zorder=5)

ax4.set_xlabel('u', fontsize=10)
ax4.set_ylabel('v', fontsize=10)
ax4.set_zlabel('Potential', fontsize=10)
ax4.set_title('Energy Landscape Analogy', fontsize=13, fontweight='bold')

# Plot 5: Effect of cooperativity
ax5 = plt.subplot(2, 3, 5)

# Test different n values
n_values = np.linspace(1.0, 3.0, 9)
num_stable_states = []

for n_val in n_values:
    test_params = params.copy()
    test_params['n'] = n_val
    
    states = find_steady_states(test_params)
    
    # Count stable ones
    stable_count = 0
    for u_val, v_val in states:
        analysis = analyze_stability(u_val, v_val, test_params)
        if analysis['stability'] == "STABLE":
            stable_count += 1
    
    num_stable_states.append(stable_count)

ax5.plot(n_values, num_stable_states, 'bo-', linewidth=2.5, markersize=8)
ax5.axvline(x=1.0, color='red', linestyle='--', alpha=0.5, label='n=1 threshold')
ax5.axhline(y=2, color='green', linestyle='--', alpha=0.5, label='Bistability')

ax5.set_xlabel('Cooperativity (n)', fontsize=11)
ax5.set_ylabel('Number of Stable States', fontsize=11)
ax5.set_title('Bistability Requires Cooperativity', fontsize=13, fontweight='bold')
ax5.grid(True, alpha=0.2)
ax5.legend(fontsize=9)
ax5.set_xticks([1.0, 1.5, 2.0, 2.5, 3.0])

# Plot 6: Switching between states
ax6 = plt.subplot(2, 3, 6)

# Simulate a switch by temporarily inhibiting one repressor
def toggle_with_induction(t, y, params):
    u_val, v_val = y
    
    # Add induction pulse: inhibits v between t=20 and t=25
    induction_strength = 10.0 if 20 < t < 25 else 0.0
    
    du_dt = params['alpha1']/(1 + v_val**params['n']) - params['beta']*u_val
    dv_dt = params['alpha2']/(1 + u_val**params['n']) - params['gamma']*v_val
    
    # Induction reduces effective concentration of v
    dv_dt = dv_dt + 0.15*induction_strength * v_val # XXX
    
    return [du_dt, dv_dt]

# Start in State 1 (high u, low v)
sol_switch = solve_ivp(
    toggle_with_induction,
    [0, 60], [10, 0.1], args=(params,),
    max_step=0.1, dense_output=True
)

ax6.plot(sol_switch.t, sol_switch.y[0], 'b-', linewidth=2.5, label='Repressor 1 (u)')
ax6.plot(sol_switch.t, sol_switch.y[1], 'r-', linewidth=2.5, label='Repressor 2 (v)')
ax6.axvspan(20, 25, alpha=0.2, color='gray', label='Induction pulse')

ax6.set_xlabel('Time', fontsize=11)
ax6.set_ylabel('Concentration', fontsize=11)
ax6.set_title('Switching States with Induction', fontsize=13, fontweight='bold')
ax6.legend(fontsize=9, loc='upper right')
ax6.grid(True, alpha=0.2)

plt.suptitle('Genetic Toggle Switch: Bistability Analysis', fontsize=16, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# 5. BIOLOGICAL SUMMARY
print("\n" + "=" * 60)
print("5. BIOLOGICAL SUMMARY: Key Insights")

print("\nA. THE THREE STEADY STATES:")
print("   1. State 1 (Stable): High u, Low v")
print("   2. State 2 (Stable): Low u, High v")  
print("   3. State 3 (Unstable): Intermediate u and v (the 'threshold')")

print("\nB. BIOLOGICAL INTERPRETATION:")
print("   • Each stable state = a possible cell fate")
print("   • The unstable state = the decision point")
print("   • Initial conditions determine final fate")
print("   • Once in a state, the system stays there (memory)")

print("\nC. SWITCHING MECHANISM:")
print("   • Transient signal temporarily inhibits one repressor")
print("   • Pushes system across the threshold (unstable state)")
print("   • System then goes to the opposite stable state")

print("\nD. KEY REQUIREMENTS FOR BISTABILITY:")
print("   1. Cooperativity (n > 1): Sharp repression needed")
print("   2. Balanced parameters: Both genes must be able to 'win'")
print("   3. Mutual inhibition: Each represses the other")

print("\n" + "=" * 60)
print("TRY MODIFYING THESE PARAMETERS TO EXPLORE:")
print("""
# Make asymmetric - one gene dominates
params['alpha1'] = 15.0  # Stronger promoter for gene 1

# Change cooperativity - affects switch sharpness  
params['n'] = 1.5  # Less cooperative, weaker bistability

# Change degradation rates - affects steady state positions
params['beta'] = 0.5  # u degrades slower
""")
````
